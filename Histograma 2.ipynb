Ecualizacion de un Histograma se decea normalizar una imagen a un punto optimo

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams ['figure.figsize']=[10,5]
imagen=cv2.imread('imagenes/original/flores.jpeg',0)
histograma ,calc =np.histogram(imagen.flatten(),256,[0,256])
distAcum=histograma.cumsum()
distAcum_normalizado=distAcum*histograma.max()/distAcum.max()
ecualizado=cv2.equalizeHist(imagen)
histograma_equalizado,_=np.histogram(ecualizado.flatten(),256,[0,256])


plt.figure(figsize=(10,5))
plt.subplot(221)
plt.title("Original")
plt.axis("off")
plt.imshow(imagen, cmap="gray")

# Histogram of the original image
plt.subplot(222)
plt.title("Histogram")
plt.axis("off")
plt.plot(distAcum_normalizado, color="b")
plt.hist(imagen.flatten(), 256, [0, 256], color="r")
plt.xlim([0, 256])
plt.legend(("Cumulative Distribution", "Histogram"), loc="upper left")

plt.subplot(223)
plt.title("Equalized")
plt.axis("off")
plt.imshow(ecualizado, cmap="gray")

plt.subplot(224)
plt.title("Equalized Histogram")
plt.axis("off")
plt.plot(histograma_equalizado, color="r")
plt.xlim([0, 256])
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [10, 5]
imagen = cv2.imread('imagenes/original/flores.jpeg')
gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
ecualizada = cv2.equalizeHist(gray)
ecu_color = cv2.merge([
    cv2.equalizeHist(imagen[:, :, 0]), 
    cv2.equalizeHist(imagen[:, :, 1]), 
    cv2.equalizeHist(imagen[:, :, 2])   
])

plt.imshow(cv2.cvtColor(ecualizada, cv2.COLOR_BGR2RGB), cmap="gray")
plt.title("Equalized Grayscale Image")
plt.show()
plt.imshow(cv2.cvtColor(ecu_color, cv2.COLOR_BGR2RGB))
plt.title("Equalized Color Image")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [10, 5]
imagen = cv2.imread('imagenes/original/flores.jpeg',0)
clip=2.0
tile=8
clahe=cv2.createCLAHE(clipLimit=clip,tileGridSize=(tile,tile))
ecualizada=clahe.apply(imagen)
plt.imshow(imagen, cmap="gray")
plt.title("imagen original")
plt.show()

plt.imshow(ecualizada, cmap="gray")
plt.title("Equalizada")
plt.show()

```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [10, 5]
imagen = cv2.imread('imagenes/original/flores.jpeg',0)
transformada=np.fft.fft2(imagen)# aplica transformada e inversa de forier

desplazar=np.fft.fftshift(transformada)#separa el espectro
magnitud_espectro=np.log(1+np.abs(desplazar))
plt.subplot(1, 3, 1)
plt.imshow(imagen, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(ecualizada, cmap='gray')
plt.title("Equalizada")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(magnitud_espectro, cmap='gray')
plt.title("Magnitud Espectro ")
plt.axis('off')

# Show the plot
plt.show()
```
