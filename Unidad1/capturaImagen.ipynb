```python
import cv2
import numpy as np
import matplotlib.pylab as plt
plt.rcParams["figure.figsize"]=[10,5]
path="imagenes/original/gato.jpeg"
img=cv2.imread(path,cv2.IMREAD_COLOR)    
imgRGB=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
r=img[:,:,2]
g=img[:,:,1]
b=img[:,:,0]

rojo_imagen=np.zeros_like(imgRGB)
rojo_imagen[:,:,0]=r

verde_imagen=np.zeros_like(imgRGB)
verde_imagen[:,:,1]=g

azul_imagen=np.zeros_like(imgRGB)
azul_imagen[:,:,2]=b

plt.subplot(141)
plt.title("canal original")
plt.imshow(imgRGB)
plt.axis("off")

plt.subplot(142)
plt.title("canal rojo")
plt.imshow(rojo_imagen)
plt.axis("off")

plt.subplot(143)
plt.title("canal verde")
plt.imshow(verde_imagen)
plt.axis("off")

plt.subplot(144)
plt.title("canal azul")
plt.imshow(azul_imagen)
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pylab as plt
plt.rcParams["figure.figsize"]=[10,5]
path="imagenes/original/gato.jpeg"
img_bgr=cv2.imread(path) 
imgRGB=cv2.cvtColor(img_bgr,cv2.COLOR_BGR2RGB) 

print("alto",img_bgr.shape[0])
print("ancho",img_bgr.shape[1])
print("canales",img_bgr.shape[2])
fig,axs=plt.subplots(1,4,figsize=(20,5))
axs[0].imshow(imgRGB)
axs[0].set_title("imagen original")
axs[0].axis("off")

fig_rojo=axs[1].imshow(img_bgr[...,0],cmap="gray")
fig.colorbar(fig_rojo,ax=axs[1],shrink=0.5)
axs[1].set_title("canal rojo")
axs[1].axis("off")

fig_azul=axs[2].imshow(img_bgr[...,1],cmap="gray")
fig.colorbar(fig_azul,ax=axs[2],shrink=0.5)
axs[2].set_title("canal azul")
axs[2].axis("off")

fig_verde=axs[3].imshow(img_bgr[...,2],cmap="gray")
fig.colorbar(fig_verde,ax=axs[3],shrink=0.5)
axs[3].set_title("canal verde")
axs[3].axis("off")


plt.show()
```
