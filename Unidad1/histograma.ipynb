umbralizaciones por defecto de tuzero en opencv

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen=cv2.imread("imagenes/original/tigre.jpeg",0)
_,umbr_trunc=cv2.threshold(imagen,127,255,cv2.THRESH_TRUNC)
_,umbr_tozero=cv2.threshold(imagen,127,255,cv2.THRESH_TOZERO)
_,umbr_tozero_invert=cv2.threshold(imagen,127,255,cv2.THRESH_TOZERO_INV)

plt.figure(figsize=(12,5))
plt.subplot(1,4,1)
plt.title("original")
plt.imshow(imagen,cmap="gray")

plt.subplot(1,4,2)
plt.title("trunc")
plt.imshow(umbr_trunc,cmap="gray")

plt.subplot(1,4,3)
plt.title("tozero")
plt.imshow(umbr_tozero,cmap="gray")

plt.subplot(1,4,4)
plt.title("tozero invertido")
plt.imshow(umbr_tozero_invert,cmap="gray")
plt.show()


```

Contraste

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def contraste(imagen,factor=1.0):
    img=imagen.astype(np.float32)
    media=np.mean(img)
    imagen_contraste=media+(img-media)*factor
    imagen_contraste=np.clip(imagen_contraste,0,255).astype(np.uint8)
    return imagen_contraste

imagen=cv2.imread("imagenes/original/tigre.jpeg",0)
factor_contrast=4
img_contraste=contraste(imagen,factor_contrast)
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.title("original")
plt.imshow(imagen,cmap="gray")
plt.subplot(1,2,2)
plt.title("imagen contraste")
plt.imshow(img_contraste,cmap="gray")

plt.show()
```

constraste a colores

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def contraste_color(imagen,factor=1.0):
    img=imagen.astype(np.float32)
    for i in range(3):
        media=np.mean(img[:,:,i])
        img[:,:,i]=media+(img[:,:,i]-media)*factor
        img[:,:,i]=np.clip(img[:,:,i],0,255)
    return img.astype(np.uint8)

imagen=cv2.imread("imagenes/original/tigre.jpeg")
factor_contrast=7
img_contraste=contraste_color(imagen,factor_contrast)
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.title("original")
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.subplot(1,2,2)
plt.title("imagen contraste")
plt.imshow(cv2.cvtColor(img_contraste,cv2.COLOR_BGR2RGB))

plt.show()
```

contraste y brillo

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def ajuste(imagen,contraste=1.0,brillo=0):
    img=imagen.astype(np.float32)
    media=np.mean(img)
    imagen_ajustada=media+(img-media)*contraste+brillo
    imagen_ajustada=np.clip(imagen_ajustada,0,255).astype(np.uint8)
    return imagen_ajustada
imagen=cv2.imread("imagenes/original/tigre.jpeg")
factor_contrast=123
factor_brillo=-123
img_ajustada=ajuste(imagen,factor_contrast,factor_brillo)
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.title("original")
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))
plt.subplot(1,2,2)
plt.title("imagen brillo")
plt.imshow(cv2.cvtColor(img_ajustada,cv2.COLOR_BGR2RGB))

plt.show()
```

histograma de una imagen con constraste

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def contraste(imagen,factor=1.0):
    img=imagen.astype(np.float32)
    media=np.mean(img)
    contrastada=media+(img-media)*factor
    contrastada=np.clip(contrastada,0,255).astype(np.uint8)
    return contrastada
imagen=cv2.imread("imagenes/original/tigre.jpeg",0)
factor_contrast=123
img_contraste=contraste(imagen,factor_contrast)
histograma_ori=cv2.calcHist([imagen],[0],None,[256],[0,256])
histograma_nueva=cv2.calcHist([img_contraste],[0],None,[256],[0,256])
plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.title("hitograma original")
plt.imshow(imagen,cmap="gray")
plt.subplot(2,2,2)
plt.title("imagen hitograma nueva")
plt.plot(histograma_ori,color="red")
plt.xlabel("intencidad")
plt.ylabel("frecuencia")




plt.subplot(2,2,3)
plt.title("imagen hitograma nueva")
plt.imshow(img_contraste,cmap="gray")

plt.subplot(2,2,4)
plt.title("imagen hitograma contraste")
plt.plot(histograma_nueva,color="red")
plt.xlabel("intencidad")
plt.ylabel("frecuencia")
plt.show()
```

mascara

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread("imagenes/original/flores.jpeg")
alto, ancho = imagen.shape[:2]
mascara = np.zeros_like(imagen)
centro = (alto//2, ancho//2)
radio = 20
cv2.circle(mascara, centro, radio, (255, 255, 255), -1)
img_mascara = cv2.bitwise_and(imagen, imagen, mask=cv2.cvtColor(mascara, cv2.COLOR_BGR2GRAY))

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(imagen)
plt.subplot(1, 2, 2)
plt.imshow(img_mascara)
plt.show()
```
