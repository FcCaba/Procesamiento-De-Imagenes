```python
import cv2,os
import numpy as np
import matplotlib.pylab as plt
```

```python
paths = ["imagenes/a jpeg/flor2.jpg", "imagenes/a jpeg/gato2.jpg", "imagenes/a jpeg/tank.jpg"]
```

Muestra De Subplot 1

```python

def mosIMGparte1(paths, formato):
    plt.figure(figsize=(15, 5))
    for i in range(len(paths)):
        print(paths[i])
        img_bgr = cv2.imread(paths[i])
        img_convertida = cv2.cvtColor(img_bgr, formato)
        plt.subplot(1, len(paths), i + 1)
        plt.imshow(img_convertida)
        plt.axis("off") 
        plt.title(f"Imagen {i+1}")
mosIMGparte1(paths, cv2.COLOR_BGR2RGB)
mosIMGparte1(paths, cv2.COLOR_RGB2BGR)
plt.show()
```

```python
def mosIMGpartegray(paths, formato):
    plt.figure(figsize=(15, 5))
    for i in range(len(paths)):
        print(paths[i])
        img_bgr = cv2.imread(paths[i])
        img_convertida = cv2.cvtColor(img_bgr, formato)
        plt.subplot(1, len(paths), i + 1)
        plt.imshow(img_convertida[...,i], cmap="gray")
        plt.axis("off") 
        plt.title(f"Imagen {i+1}")
mosIMGpartegray(paths, cv2.COLOR_BGR2RGB)
plt.show()
```

```python
listaCmap=["Reds","Greens","Blues"]
def mosIMGparte1(paths, formato):
    plt.figure(figsize=(15, 5))
    for i in range(len(paths)):
        print(paths[i])
        img = cv2.imread(paths[i])
        img_convertida = cv2.cvtColor(img, formato)
        plt.subplot(1, len(paths), i + 1)
        plt.imshow(cv2.split(img)[i],cmap=listaCmap[i])
        plt.axis("off") 
        plt.title(f"Imagen {i+1}")
mosIMGparte1(paths, cv2.IMREAD_COLOR)
plt.show()
```

Muestra De colorbar 2

```python

def mosIMGparte2(paths, formato):
    plt.figure(figsize=(15, 5))
    fig,axs=plt.subplots(1,len(paths),figsize=(20,5))
    for i in range(len(paths)):
        print(paths[i])
        img_bgr = cv2.imread(paths[i])
        img_convertida = cv2.cvtColor(img_bgr, formato)
        imgFIG = axs[i].imshow(img_convertida)
        fig.colorbar(imgFIG, ax=axs[i],shrink=0.5)
        plt.axis("off") 
        plt.title(f"Imagen {i+1}")
mosIMGparte2(paths, cv2.COLOR_BGR2RGB)
mosIMGparte2(paths, cv2.COLOR_RGB2BGR)
plt.show()
```

```python
def mosIMGpartegray(paths, formato):
    plt.figure(figsize=(15, 5))
    fig,axs=plt.subplots(1,len(paths),figsize=(20,5))
    for i in range(len(paths)):
        print(paths[i])
        img_bgr = cv2.imread(paths[i])
        img_convertida = cv2.cvtColor(img_bgr, formato)
        imgFIG = axs[i].imshow(img_convertida[...,i], cmap="gray")
        fig.colorbar(imgFIG, ax=axs[i],shrink=0.5)
        plt.axis("off") 
        plt.title(f"Imagen {i+1}")
mosIMGpartegray(paths, cv2.COLOR_BGR2RGB)

plt.show()
```

Funciones de Guardado Forma Normal

```python

def GuardarIMGparte1(paths, formato,titulo):
    plt.figure(figsize=(15, 5))
    for i in range(len(paths)):
        #print(paths[i])
        img_bgr = cv2.imread(paths[i])
        img_convertida = cv2.cvtColor(img_bgr, formato)
        if os.path.exists(f"imagenes/Foto Persona/Forma1/{titulo}") == False:
            os.makedirs(f"imagenes/Foto Persona/Forma1/{titulo}")
        else:
            plt.imsave(f"imagenes/Foto Persona/Forma1/{titulo}/img_{i}{titulo}.jpg",cv2.cvtColor(img_convertida,cv2.COLOR_BGR2RGB))
GuardarIMGparte1(paths, cv2.COLOR_BGR2RGB,"bgr")
GuardarIMGparte1(paths, cv2.COLOR_RGB2BGR,"rgb")
```

```python
def GuardarIMGgray(paths, formato, titulo):
    plt.figure(figsize=(15, 5))
    for i in range(len(paths)):
        img_bgr = cv2.imread(paths[i])
        img_gray = cv2.cvtColor(img_bgr, formato)
        nomIMG=f"img_{i}_{titulo}.jpg"
        if not os.path.exists(f"imagenes/Foto Persona/Forma1/{titulo}"):
            os.makedirs(f"imagenes/Foto Persona/Forma1/{titulo}")
        else:
            if os.path.exists(f"imagenes/Foto Persona/Forma1/{titulo}") == False:
                plt.imsave(f"imagenes/Foto Persona/Forma1/{titulo}/{nomIMG}", img_gray, cmap='gray')
                print(f"Imagen guardada en: imagenes/Foto Persona/Forma1/{titulo}/{nomIMG}")
            else:
                print("La carpeta ya existe")
GuardarIMGgray(paths, cv2.COLOR_BGR2GRAY, "gris")
```

Funciones de Guardado Forma colorbar

```python

def guardarIMGparte2(paths, formato,titulo):
    plt.figure(figsize=(15, 5))
    fig,axs=plt.subplots(1,len(paths),figsize=(20,5))
    for i in range(len(paths)):
        print(paths[i])
        img_bgr = cv2.imread(paths[i])
        if titulo=="gris":
            img_convertida = cv2.cvtColor(img_bgr, formato)
            imgFIG = axs[i].imshow(img_convertida, cmap="gray")
        else:
            img_convertida = cv2.cvtColor(img_bgr, formato)
            imgFIG = axs[i].imshow(img_convertida)
        fig.colorbar(imgFIG, ax=axs[i],shrink=0.5)
        plt.axis("off") 
        plt.title(f"Imagen {i+1}")
        nomIMG=f"img_{i}_{titulo}.jpg"
        if not os.path.exists(f"imagenes/Foto Persona/Forma2/{titulo}"):
            os.makedirs(f"imagenes/Foto Persona/Forma2/{titulo}")
        else:
            if os.path.exists(f"imagenes/Foto Persona/Forma2/{nomIMG}") == False:
                print("entro")
                if titulo=="gris":
                    plt.imsave(f"imagenes/Foto Persona/Forma2/{titulo}/{nomIMG}", img_convertida, cmap='gray')
                else:
                    plt.imsave(f"imagenes/Foto Persona/Forma2/{titulo}/{nomIMG}", img_convertida)
                print(f"Imagen guardada en: imagenes/Foto Persona/Forma2/{titulo}/{nomIMG}")
            else:
                print("La imagen ya existe")
guardarIMGparte2(paths, cv2.COLOR_BGR2RGB,"bgr")
guardarIMGparte2(paths, cv2.COLOR_RGB2BGR,"rpg")
guardarIMGparte2(paths, cv2.COLOR_BGR2GRAY,"gris")
plt.show()
```
