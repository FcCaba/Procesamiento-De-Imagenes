Canales reales 

```python
import cv2
import numpy as np
import matplotlib.pylab as plt
plt.rcParams["figure.figsize"]=[10,5]
path="imagenes/original/gato.jpeg"
img=cv2.imread(path,cv2.IMREAD_COLOR)    
imgRGB=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
r=img[:,:,2]
g=img[:,:,1]
b=img[:,:,0]

rojo_imagen=np.zeros_like(imgRGB)
rojo_imagen[:,:,0]=r

verde_imagen=np.zeros_like(imgRGB)
verde_imagen[:,:,1]=g

azul_imagen=np.zeros_like(imgRGB)
azul_imagen[:,:,2]=b

plt.subplot(141)
plt.title("canal original")
plt.imshow(imgRGB)
plt.axis("off")

plt.subplot(142)
plt.title("canal rojo")
plt.imshow(rojo_imagen)
plt.axis("off")

plt.subplot(143)
plt.title("canal verde")
plt.imshow(verde_imagen)
plt.axis("off")

plt.subplot(144)
plt.title("canal azul")
plt.imshow(azul_imagen)
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pylab as plt
plt.rcParams["figure.figsize"]=[10,5]
path="imagenes/original/gato.jpeg"
img=cv2.imread(path,cv2.IMREAD_COLOR)    
b,g,r=cv2.split(img)

plt.subplot(141)
plt.title("original")
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))

plt.subplot(142)
plt.title("canal rojo")
plt.imshow(r,cmap="Reds")
plt.axis("off")

plt.subplot(143)
plt.title("canal verde")
plt.imshow(g,cmap="Greens")
plt.axis("off")

plt.subplot(144)
plt.title("canal azul")
plt.imshow(b,cmap="Blues")
plt.axis("off")

plt.show
```
