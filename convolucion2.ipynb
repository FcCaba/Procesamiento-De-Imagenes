```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def convolucion_media_horizontal(imagen,tama_kernel):
    kernel=np.ones((1,tama_kernel),np.float32)/tama_kernel
    img_final=cv2.filter2D(imagen,-1,kernel)
    return img_final
img=cv2.imread(r'imagenes/original/distoricion horizontal 2.jpg')
gris=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

imgres=convolucion_media_horizontal(gris,5)

plt.figure(figsize=(15,5))
plt.subplot(121)
plt.title("original")
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))

plt.subplot(122)
plt.title("filtro")
plt.imshow(imgres,cmap="gray")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def convolucion_media_diagonal(imagen,tama_kernel):
    kernel = np.ones((tama_kernel, tama_kernel), np.float32) / (tama_kernel * tama_kernel)
    img_final=cv2.filter2D(imagen,-1,kernel)
    return img_final
img=cv2.imread(r'imagenes/img distorcion/diagonal.jpg')
rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
imgresR=convolucion_media_diagonal(rgb[:,:,0],174)
imgresG=convolucion_media_diagonal(rgb[:,:,1],174)
imgresB=convolucion_media_diagonal(rgb[:,:,2],174)

plt.figure(figsize=(25,10))
plt.subplot(141)
plt.title("original")
plt.axis("off")
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))

plt.subplot(142)
plt.title("filtro R")
plt.axis("off")
plt.imshow(imgresR,cmap="gray")

plt.subplot(143)
plt.title("filtro G")
plt.axis("off")
plt.imshow(imgresG,cmap="gray")

plt.subplot(144)
plt.title("filtro B")
plt.imshow(imgresB,cmap="gray")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def filtro_gaussiano(imagen, kernel_size):
    return cv2.GaussianBlur(imagen, (kernel_size, kernel_size), 0)

img=cv2.imread(r'imagenes/img distorcion/dis2.jfif')
rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
imgresR=filtro_gaussiano(rgb[:,:,0],15)
imgresG=filtro_gaussiano(rgb[:,:,1],15)
imgresB=filtro_gaussiano(rgb[:,:,2],15)

plt.figure(figsize=(25,10))
plt.subplot(141)
plt.title("original")
plt.axis("off")
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))

plt.subplot(142)
plt.title("filtro R")
plt.axis("off")
plt.imshow(imgresR,cmap="gray")

plt.subplot(143)
plt.title("filtro G")
plt.axis("off")
plt.imshow(imgresG,cmap="gray")

plt.subplot(144)
plt.title("filtro B")
plt.axis("off")
plt.imshow(imgresB,cmap="gray")

plt.show()
```
