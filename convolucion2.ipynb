```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def convolucion_media_horizontal(imagen,tama_kernel):
    kernel=np.ones((1,tama_kernel),np.float32)/tama_kernel
    img_final=cv2.filter2D(imagen,-1,kernel)
    return img_final
img=cv2.imread(r'imagenes/original/distoricion horizontal 2.jpg')
gris=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)

imgres=convolucion_media_horizontal(gris,5)

plt.figure(figsize=(15,5))
plt.subplot(121)
plt.title("original")
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))

plt.subplot(122)
plt.title("filtro")
plt.imshow(imgres,cmap="gray")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def convolucion_media_horizontal(imagen,tama_kernel):
    kernel=np.ones((1,tama_kernel),np.float32)/tama_kernel
    img_final=cv2.filter2D(imagen,-1,kernel)
    return img_final
img=cv2.imread(r'imagenes/original/distoricion horizontal.jpg')
rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
imgresR=convolucion_media_horizontal(rgb[:,:,0],25)
imgresG=convolucion_media_horizontal(rgb[:,:,1],25)
imgresB=convolucion_media_horizontal(rgb[:,:,2],25)

plt.figure(figsize=(25,10))
plt.subplot(141)
plt.title("original")
plt.axis("off")
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))

plt.subplot(142)
plt.title("filtro R")
plt.axis("off")
plt.imshow(imgresR,cmap="gray")

plt.subplot(143)
plt.title("filtro G")
plt.axis("off")
plt.imshow(imgresG,cmap="gray")

plt.subplot(144)
plt.title("filtro B")
plt.imshow(imgresB,cmap="gray")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def convolucion_media_diagonal(imagen,tama_kernel):
    kernel = np.ones((tama_kernel, tama_kernel), np.float32) / (tama_kernel * tama_kernel)
    img_final=cv2.filter2D(imagen,-1,kernel)
    return img_final
img=cv2.imread(r'imagenes/original/distoricion.jpg')
rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
imgresR=convolucion_media_diagonal(rgb[:,:,0],25)
imgresG=convolucion_media_diagonal(rgb[:,:,1],25)
imgresB=convolucion_media_diagonal(rgb[:,:,2],25)

plt.figure(figsize=(25,10))
plt.subplot(141)
plt.title("original")
plt.axis("off")
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))

plt.subplot(142)
plt.title("filtro R")
plt.axis("off")
plt.imshow(imgresR,cmap="gray")

plt.subplot(143)
plt.title("filtro G")
plt.axis("off")
plt.imshow(imgresG,cmap="gray")

plt.subplot(144)
plt.title("filtro B")
plt.imshow(imgresB,cmap="gray")

plt.show()
```
