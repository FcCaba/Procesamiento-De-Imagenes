Ejercicio 1
Desarrolle una aplicación que calcule el factorial de un numero entero introducido por 
teclado  

```python
def calcular_factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        factorial = 1
        for i in range(1, n + 1):
            factorial *= i
        return factorial
numero = int(input("ingrese un numero:"))
print("El factorial de", numero, "es:", calcular_factorial(numero))
```

2.- Desarrolle una aplicación que determine si un numero introducido por teclado es primo o 
no  

```python
def es_primo(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
numero = int(input("ingrese un numero:"))
if es_primo(numero):
    print(numero, "es primo.")
else:
    print(numero, "no es primo.")
```

3.- Desarrolle una aplicación que permita ingresar diferentes números por teclado con un limite 
definido inicialmente (no menor a 3 números) y determine cual de ellos es el mayor y el menor.

```python
import os
numerolist = []
while True:
    os.system("cls")
    n = int(input("Ingrese los números necesarios para calcular.\n0 para calcular: "))
    if n == 0:
        if len(numerolist) >= 3:
            break
        else:
            print("Ingrese más de 3 valores")
    else:
        numerolist.append(n)

print(numerolist)
numerolist.sort()
print("numero menor",numerolist[0])
print("numero mayor:",numerolist[-1])
```

4.-Desarrolle una aplicación que permita identificar si una cadena introducida por teclado es 
un palíndromo ( oso es palíndromo )

```python
def es_palindromo(tex):
    tex = tex.replace(" ", "").lower()  
    return tex == tex[::-1] 
texto = input("Introduce una palabra:")
if es_palindromo(texto):
    print(f"'{texto}' es un palíndromo.")
else:
    print(f"'{texto}' no es un palíndromo.")
```

5.- desarrolle una aplicación que permita introducir 7 números por teclado y determine cual de 
ellos se repite más veces y cuantas.

```python
from collections import Counter
import os
os.system("cls")
def numero_que_mas_se_repite():
    numeros = []
    for i in range(7):
        num = int(input(f"ingrese un numero {i+1}: "))
        numeros.append(num)
    
    contador = Counter(numeros)
    numComun = contador.most_common(1)[0] 
    print(numeros)
    print(f"numero que se repite {numComun[0]} y se repite {numComun[1]} veces.")

numero_que_mas_se_repite()
```
