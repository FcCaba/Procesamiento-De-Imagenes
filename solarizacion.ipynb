Solarizacin

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
def solarizar(imagen,umbral):
    imagen_solarizada=np.where(imagen<umbral,imagen,255-imagen)
    return imagen_solarizada
imagen = cv2.imread('imagenes/original/tigre.jpeg')
img_gris=cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)
umbral=127
img_nueva=solarizar(img_gris,umbral)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("original")
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))

plt.subplot(1,2,2)
plt.title("solarizada")
plt.imshow(img_nueva,cmap="gray")

plt.show()
```

Posteriazada

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
def posterizada(imagen,nivel):
    factor=255/nivel
    img_post=np.round((imagen/factor)*factor)
    
    return img_post.astype(np.uint8)
imagen = cv2.imread('imagenes/original/tigre.jpeg')
img_gris=cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)
nivel=4
img_nueva=posterizada(img_gris,nivel)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("original")
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))

plt.subplot(1,2,2)
plt.title("posterizada")
plt.imshow(img_nueva,cmap="gray")

plt.show()
```

Con Falso Color

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
def falso_color(imagen_Gray):
    escalar_imagen=cv2.normalize(imagen_Gray,None,0,255,cv2.NORM_MINMAX)
    apli_color=cv2.applyColorMap(escalar_imagen,cv2.COLORMAP_PLASMA)
    return apli_color
imagen = cv2.imread('imagenes/original/tigre.jpeg')
img_nueva=falso_color(imagen)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.title("original")
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))

plt.subplot(1,2,2)
plt.title("falso_color")
plt.imshow(cv2.cvtColor(img_nueva,cv2.COLOR_BGR2RGB))

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def procesar_imagenes2(imagen_ruta):
    imagen_gray = cv2.imread(imagen_ruta, cv2.IMREAD_GRAYSCALE)
    imagen_gray = cv2.normalize(imagen_gray, None, 0, 255, cv2.NORM_MINMAX)
    colormaps = [name for name in dir(cv2) if name.startswith("COLORMAP_")]
    n=len(colormaps)
    print(n)
    f=n//4+1
    col=4
    plt.figure(figsize=(12, 3 * f))
    plt.subplot(f, col, 1)
    plt.imshow(imagen_gray, cmap="gray")
    plt.title("Original (Gris)")
    plt.axis("off")
    cont=2
    for i,cmap in enumerate(colormaps):
        img_colored = cv2.applyColorMap(imagen_gray,i)
        img_colored = cv2.cvtColor(img_colored, cv2.COLOR_BGR2RGB)
        plt.subplot(f, col, cont)
        plt.imshow(img_colored)
        plt.title(cmap)
        plt.axis("off")
        cont+=1
        print(cmap)
procesar_imagenes2("imagenes/original/tigre.jpeg")
```
