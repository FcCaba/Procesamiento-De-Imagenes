Funciones filtros

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
def Enfoque(img):
    mascara=np.array([[0,-1,0,],
                      [-1,5,-1,],
                      [0,-1,0,]])
    img_final=cv2.filter2D(img,-1,mascara)
    return img_final
def Decenfoque(img):
    mascara=np.array([[1,1,1,],
                  [1,1,1,],
                  [1,1,1,]])
    img_final=cv2.filter2D(img,-1,mascara)
    return img_final
def Realce_bordes(img):
    mascara=np.array([[0,0,0,],
                  [-1,1,0,],
                  [0,0,0,]])
    img_final=cv2.filter2D(img,-1,mascara)
    return img_final
def Repajado(img):
    mascara=np.array([[-2,-1,0,],
                  [-1,1,1,],
                  [0,1,2,]])
    img_final=cv2.filter2D(img,-1,mascara)
    return img_final
def DetencionBordes(img):
    mascara=np.array([[0,-1,0,],
                  [-1,5,-1,],
                  [0,-1,0,]])
    img_final=cv2.filter2D(img,-1,mascara)
    return img_final
def Sobel(img):
    mascara=np.array([[-1,0,1,],
                  [-2,0,2,],
                  [-1,0,1,]])
    img_final=cv2.filter2D(img,-1,mascara)
    return img_final
def Sharpen(img):
    mascara=np.array([[1,-2,1,],
                  [-2,5,-2,],
                  [1,-2,1,]])
    img_final=cv2.filter2D(img,-1,mascara)
    return img_final
def FiltroNorte(img):
    mascara=np.array([[1,1,1,],
                  [1,-2,1,],
                  [-1,-1,-1,]])
    img_final=cv2.filter2D(img,-1,mascara)
    return img_final
def FiltroEste(img):
    mascara=np.array([[-1,1,1,],
                  [-1,-2,1,],
                  [-1,1,1,]])
    img_final=cv2.filter2D(img,-1,mascara)
    return img_final
def Gauss(img):
    mascara=np.array([[1,2,3,1,1],
                  [2,7,11,7,2],
                  [3,11,17,11,3],
                  [2,7,11,7,1],
                  [1,2,3,2,1],])
    img_final=cv2.filter2D(img,-1,mascara)
    return img_final
img=cv2.imread('D:\Franz Caballero\Procesamiento de imagenes\Actividades\Ejercicios python\imagenes\original\gato.jpeg')
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
enfoque=Enfoque(img)
decenfoque=Decenfoque(img)
realce=Realce_bordes(img)
repajado=Repajado(img)
detencion=DetencionBordes(img)
sobel=Sobel(img)
sharpen=Sharpen(img)
norte=FiltroNorte(img)
este=FiltroEste(img)
gauss=Gauss(img)
filtros=[enfoque,decenfoque,realce,repajado,detencion,sobel,sharpen,norte,este,gauss]
plt.figure(figsize=(15,10))
for i, f in enumerate(filtros):
    plt.subplot(1, 10, i+1)  
    plt.title(f'Filtro {i+1}')
    plt.axis('off')
    plt.imshow(f)
plt.show()
```
