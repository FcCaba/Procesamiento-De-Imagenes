```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img=cv2.imread('D:\Franz Caballero\Procesamiento de imagenes\Actividades\Ejercicios python\imagenes\original\gato.jpeg')
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
kernel5x5=np.ones((5,5),np.float32)/25 #suavisado
img_suavisado1=cv2.filter2D(img,-1,kernel5x5)
kernel7x7=np.ones((7,7),np.float32)/49 #suavisado
img_suavisado2=cv2.filter2D(img,-1,kernel7x7)
plt.figure(figsize=(10,5))
plt.subplot(131)
plt.title('Imagen Original')
plt.axis('off')
plt.imshow(img)

plt.subplot(132)
plt.title('Imagen con mascara 1')
plt.axis('off')
plt.imshow(img_suavisado1,cmap='gray')

plt.subplot(133)
plt.title('Imagen con mascara 2')
plt.axis('off')
plt.imshow(img_suavisado2,cmap='gray')
plt.show()
```
